```matlab
clc;
clear;
close all;

[m, fs] = audioread("audio2.aac")
```

```matlab
sound(m, fs);

[s, ch] = size(m)
```

```matlab
t = (0:(s-1))/fs
```

```matlab

figure;
subplot(2, 1, 1)
% Plot the waveform
plot(m);
title('Waveform of Imported Audio');
xlabel('Sample Number');
ylabel('Amplitude');
subplot(2, 1, 2)
plot(t, m);
title("waveform of signal on time axis");
```

```matlab

msg = m(:, 1) % original message signal we will use
```

```matlab
msg = msg/max(abs(msg));  % to make it in range of 1 to -1 for quantization
N = length(msg);
t = (0: N-1)/fs
```

```matlab
figure;
subplot(2, 1, 1)
plot(t, m(:, 1))
title("Mono")
```

```matlab
%% Quantization
x = msg;

bits = 4;
L = 2^bits;

xmin = -1;
xmax = 1;

delta = (xmax-xmin)/L;

disp(delta);
```

```matlab

xq = floor((x-xmin)/delta);

disp(xq);
```

```matlab

% Limit the maximum index
xq(xq >= L) = L-1;

%% PCM Encoding
pcm_mod = dec2bin(xq, bits);

%% Display
disp('Quantization indices:');
```

```matlab
disp(xq);
```

```matlab

disp('PCM codes:');
```

```matlab
disp(pcm_mod);
```

```matlab

pcm_bits = reshape(pcm_mod.', 1, []);
disp('Serial PCM bit stream:');
```

```matlab
disp(pcm_bits);
```

```matlab
%% Convert characters to numerical bits
pcm_bits_num = pcm_bits - '0';  % convert to numeric values by subtracting '0' or (48 in ascii value).

%% Plot PCM bit stream
figure;

stairs(pcm_bits_num, 'LineWidth', 1.5);

ylim([-0.2 1.2]);
xlabel('Bit number');
ylabel('Amplitude');
title('PCM Bit Stream');
grid on;
```

```matlab
%% PCM Decoder

% Convert serial stream back into groups of 'bits'
num_samples = length(xq);
disp(num_samples);
```

```matlab
disp(pcm_bits);
```

```matlab

pcm_matrix = reshape(pcm_bits, bits, num_samples).';
disp(pcm_matrix);
```

```matlab

% Binary → decimal
decoded_index = bin2dec(pcm_matrix);
disp(decoded_index);
```

```matlab

%% Convert quantization index back to amplitude

% PCM scheme is mid rise quantization. so we add 0.5 but if it is mid traid
% so we do not need to add 0.5 bcz it starts from center.
xq_received = xmin + (decoded_index + 0.5)*delta; % .5 to reduce the quant. error by delta/2 from delta, in mid rise type
disp(xq_received);
```

```matlab

%% Plot original and reconstructed signal

figure;

plot(t, x, 'LineWidth', 1.2);
hold on;

stairs(t, xq_received, 'LineWidth', 1.2);

xlabel('Time (s)');
ylabel('Amplitude');
title('PCM: Original vs Reconstructed Signal');
legend('Original','Decoded');
grid on;
```

```matlab
% quantization error
xorginal = x;
eq = xorginal - xq_received;
%% Quantization error
figure;

plot(t, eq, 'LineWidth', 1.2);

xlabel('Time (s)');
ylabel('Error Amplitude');
title('Quantization Noise');
grid on;
```
